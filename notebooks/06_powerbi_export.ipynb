```python
%sql
DESCRIBE TABLE default.gold_overall_daily_activity;
```

```python
%sql
SELECT *
FROM default.gold_overall_daily_activity
ORDER BY activity_date;
```

```python
%sql
SELECT
    COUNT(*) AS row_count,
    MIN(activity_date) AS start_date,
    MAX(activity_date) AS end_date
FROM default.gold_overall_daily_activity;
```

```python
%sql
SELECT
    SUM(CASE WHEN activity_date IS NULL THEN 1 ELSE 0 END) AS null_activity_date,
    SUM(CASE WHEN total_session_count IS NULL THEN 1 ELSE 0 END) AS null_session_count,
    SUM(CASE WHEN active_station_count IS NULL THEN 1 ELSE 0 END) AS null_station_count,
    SUM(CASE WHEN active_charger_count IS NULL THEN 1 ELSE 0 END) AS null_charger_count,
    SUM(CASE WHEN total_energy_kwh IS NULL THEN 1 ELSE 0 END) AS null_energy,
    SUM(CASE WHEN estimated_revenue_inr IS NULL THEN 1 ELSE 0 END) AS null_revenue
FROM default.gold_overall_daily_activity;
```

```python
%sql
SELECT
    SUM(total_session_count) AS total_sessions,
    SUM(active_station_count) AS station_activity_sum,
    SUM(active_charger_count) AS charger_activity_sum,
    SUM(total_energy_kwh) AS total_energy_kwh,
    AVG(average_energy_kwh) AS avg_energy_kwh,
    AVG(average_duration_minutes) AS avg_duration_minutes,
    SUM(total_occupied_minutes) AS total_occupied_minutes,
    SUM(estimated_revenue_inr) AS estimated_revenue_inr
FROM default.gold_overall_daily_activity;
```

```python
%sql
DESCRIBE TABLE default.gold_station_daily_activity;
```

```python
%sql
SELECT
    COUNT(*) AS row_count,
    MIN(activity_date) AS start_date,
    MAX(activity_date) AS end_date
FROM default.gold_station_daily_activity;
```

```python
%sql
SELECT
    SUM(CASE WHEN station_id IS NULL THEN 1 ELSE 0 END) AS null_station_id,
    SUM(CASE WHEN station_name IS NULL THEN 1 ELSE 0 END) AS null_station_name,
    SUM(CASE WHEN activity_date IS NULL THEN 1 ELSE 0 END) AS null_activity_date,
    SUM(CASE WHEN daily_session_count IS NULL THEN 1 ELSE 0 END) AS null_session_count,
    SUM(CASE WHEN daily_energy_kwh IS NULL THEN 1 ELSE 0 END) AS null_energy,
    SUM(CASE WHEN daily_estimated_revenue_inr IS NULL THEN 1 ELSE 0 END) AS null_revenue
FROM default.gold_station_daily_activity;
```

```python
%sql
SELECT
    SUM(daily_session_count) AS total_sessions,
    SUM(daily_completed_session_count) AS completed_sessions,
    SUM(daily_energy_kwh) AS total_energy_kwh,
    AVG(average_session_energy_kwh) AS avg_session_energy_kwh,
    AVG(average_duration_minutes) AS avg_duration_minutes,
    SUM(daily_estimated_revenue_inr) AS estimated_revenue_inr
FROM default.gold_station_daily_activity;
```

```python
# Week 08 - Gold Source Register

gold_sources = [
    {
        "table": "default.gold_overall_daily_activity",
        "grain": "one row per activity_date",
        "purpose": "Overall EV charging network performance",
        "powerbi_use": "Overall KPIs and daily network trend"
    },
    {
        "table": "default.gold_station_daily_activity",
        "grain": "one row per station_id + activity_date",
        "purpose": "Daily station-level charging performance",
        "powerbi_use": "Station performance and station comparison"
    }
]

for source in gold_sources:
    print(f"Table: {source['table']}")
    print(f"Grain: {source['grain']}")
    print(f"Purpose: {source['purpose']}")
    print(f"Power BI use: {source['powerbi_use']}")
    print("-" * 60)
```

```python
# Week 08 - Controlled Export 1
# Source: gold_overall_daily_activity

overall_export = spark.sql("""
    SELECT
        activity_date,
        total_session_count,
        active_station_count,
        active_charger_count,
        total_energy_kwh,
        average_energy_kwh,
        average_duration_minutes,
        total_occupied_minutes,
        estimated_revenue_inr
    FROM default.gold_overall_daily_activity
    ORDER BY activity_date
""")

display(overall_export)
```

```python
# Week 08 - Controlled Export 2
# Source: gold_station_daily_activity

station_export = spark.sql("""
    SELECT
        station_id,
        station_name,
        city_band,
        zone,
        site_type,
        activity_date,
        daily_session_count,
        daily_completed_session_count,
        daily_energy_kwh,
        average_session_energy_kwh,
        average_duration_minutes,
        daily_estimated_revenue_inr
    FROM default.gold_station_daily_activity
    ORDER BY station_id, activity_date
""")

display(station_export)
```

```python
print("Station export row count:", station_export.count())
```

```python
gold_station_count = spark.sql("""
    SELECT COUNT(*) AS row_count
    FROM default.gold_station_daily_activity
""").collect()[0]["row_count"]

export_count = station_export.count()

print("Gold table rows   :", gold_station_count)
print("Export rows       :", export_count)
print("Rows match        :", gold_station_count == export_count)
```

```python
display(
    station_export.select(
        "station_id",
        "station_name",
        "activity_date"
    ).where(
        "station_id IS NULL OR station_name IS NULL OR activity_date IS NULL"
    )
)
```

```python
display(
    station_export.selectExpr(
        "MIN(activity_date) AS start_date",
        "MAX(activity_date) AS end_date"
    )
)
```

```python
display(
    station_export.groupBy(
        "station_id",
        "activity_date"
    )
    .count()
    .where("count > 1")
)
```

```python
gold_overall_total = spark.sql("""
    SELECT
        SUM(total_session_count) AS total_sessions,
        SUM(total_energy_kwh) AS total_energy,
        SUM(estimated_revenue_inr) AS total_revenue
    FROM default.gold_overall_daily_activity
""").collect()[0]

export_overall_total = overall_readback.selectExpr(
    "SUM(CAST(total_session_count AS DOUBLE)) AS total_sessions",
    "SUM(CAST(total_energy_kwh AS DOUBLE)) AS total_energy",
    "SUM(CAST(estimated_revenue_inr AS DOUBLE)) AS total_revenue"
).collect()[0]

print("Gold totals:")
print(gold_overall_total)

print("\nExport totals:")
print(export_overall_total)
```

```python
gold_station_total = spark.sql("""
    SELECT
        SUM(daily_session_count) AS total_sessions,
        SUM(daily_energy_kwh) AS total_energy,
        SUM(daily_estimated_revenue_inr) AS total_revenue
    FROM default.gold_station_daily_activity
""").collect()[0]

export_station_total = station_readback.selectExpr(
    "SUM(CAST(daily_session_count AS DOUBLE)) AS total_sessions",
    "SUM(CAST(daily_energy_kwh AS DOUBLE)) AS total_energy",
    "SUM(CAST(daily_estimated_revenue_inr AS DOUBLE)) AS total_revenue"
).collect()[0]

print("Gold totals:")
print(gold_station_total)

print("\nExport totals:")
print(export_station_total)
```
