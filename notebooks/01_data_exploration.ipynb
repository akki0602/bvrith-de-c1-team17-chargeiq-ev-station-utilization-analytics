## Mission

```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.getOrCreate()
volume_path="/Volumes/ev-charge/default/ev-data"
```

## Explore `shipments.parquet`

```python
import pandas as pd
pdf_shipments = pd.read_parquet("/Volumes/ev-charge/default/ev-data/sessions.parquet")
df_shipments = spark.createDataFrame(pdf_shipments)
display(df_shipments)
df_shipments.printSchema()
df_shipments.createOrReplaceTempView("shipments")
```

```python
spark.sql('SELECT * FROM shipments LIMIT 10').show()
```

## Explore `hubs.json`

```python
df_hubs=spark.read.format("json")\
    \
    .load("/Volumes/ev-charge/default/ev-data/chargers.json")
display(df_hubs)
df_hubs.printSchema()
df_hubs.createOrReplaceTempView("hubs")
```

## Business Key Validation

```python
%sql
-- Validate business key uniqueness
SELECT COUNT(*) physical_rows,
COUNT(DISTINCT physical_record_id) distinct_keys
FROM shipments
```

## Bronze Demonstration Table

```python
%sql
CREATE OR REPLACE TABLE workspace.default.shiptrack_week03_bronze_demo_shipments AS
SELECT * FROM shipments;
```

```python
%sql
DESCRIBE DETAIL workspace.default.shiptrack_week03_bronze_demo_shipments;
DESCRIBE HISTORY workspace.default.shiptrack_week03_bronze_demo_shipments;
```

## Lineage Demonstration View

```python
%sql
CREATE OR REPLACE VIEW workspace.default.shiptrack_week03_lineage_demo_view AS
SELECT physical_record_id, final_status FROM shiptrack_week03_bronze_demo_shipments;
```
